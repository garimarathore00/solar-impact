```python
import pandas as pd
import plotly.express as px
import os

# 📁 Folder create
os.makedirs("static/plots", exist_ok=True)

# 📊 Load Data
df = pd.read_csv("data.csv")

# 🔥 Clean columns
df.columns = df.columns.str.strip().str.lower()

# 🔥 Date convert
df['date'] = pd.to_datetime(df['date'], errors='coerce')

# ===============================
# 🔷 BASIC PREP
# ===============================
df['year'] = df['date'].dt.year
df['month'] = df['date'].dt.month
df['hour'] = df['date'].dt.hour

# ===============================
# 🔷 SPEED CATEGORY
# ===============================
def speed_cat(x):
    if x < 500:
        return "Slow"
    elif x < 1000:
        return "Medium"
    else:
        return "Fast"

df['speed_category'] = df['speed'].fillna(0).apply(speed_cat)

# ===============================
# 🔷 DIRECTION BIN
# ===============================
df['direction_bin'] = pd.cut(
    df['direction'],
    bins=[0,90,180,270,360],
    labels=['0-90','90-180','180-270','270-360']
)

# ===============================
# 🎨 STYLE FUNCTION
# ===============================
def style(fig):
    fig.update_layout(
        template="plotly_dark",
        paper_bgcolor="#000814",
        plot_bgcolor="#001d3d",
        font=dict(color="white"),
        title_x=0.5
    )
    return fig

# ===============================
# 🔷 1–10 OVERVIEW
# ===============================

style(px.line(df.groupby('year').size().reset_index(name='count'),
              x='year', y='count', title="Events Over Time"))\
.write_html("static/plots/1.html")

style(px.pie(df, names='cme_type', title="CME Type", hole=0.4))\
.write_html("static/plots/2.html")

style(px.bar(df, x='speed_category', title="Speed Category"))\
.write_html("static/plots/3.html")

style(px.histogram(df, x='speed', nbins=15, title="Speed Distribution"))\
.write_html("static/plots/4.html")

style(px.box(df, y='speed', title="Speed Box"))\
.write_html("static/plots/5.html")

style(px.bar(df, x='event_type', title="Event Type"))\
.write_html("static/plots/6.html")

style(px.bar(df, x='source_location', title="Source Location"))\
.write_html("static/plots/7.html")

style(px.bar(df, x='measurement_technique', title="Technique"))\
.write_html("static/plots/8.html")

style(px.bar(df, x='catalog', title="Catalog"))\
.write_html("static/plots/9.html")

style(px.scatter(df, x='speed', y='half_angle', title="Speed vs Angle"))\
.write_html("static/plots/10.html")

# ===============================
# 🔷 11–20 TIME
# ===============================

style(px.bar(df.groupby('month').size().reset_index(name='count'),
             x='month', y='count', title="Month"))\
.write_html("static/plots/11.html")

style(px.line(df.groupby('hour').size().reset_index(name='count'),
              x='hour', y='count', title="Hour"))\
.write_html("static/plots/12.html")

style(px.area(df.groupby('year').size().reset_index(name='count'),
              x='year', y='count', title="Year Trend"))\
.write_html("static/plots/13.html")

style(px.bar(df.groupby('quarter').size().reset_index(name='count'),
             x='quarter', y='count', title="Quarter"))\
.write_html("static/plots/14.html")

style(px.histogram(df, x='hour', title="Hour Dist"))\
.write_html("static/plots/15.html")

style(px.histogram(df, x='month', title="Month Dist"))\
.write_html("static/plots/16.html")

style(px.box(df, x='month', y='speed', title="Speed vs Month"))\
.write_html("static/plots/17.html")

style(px.box(df, x='hour', y='speed', title="Speed vs Hour"))\
.write_html("static/plots/18.html")

style(px.line(df.groupby('day').size().reset_index(name='count'),
              x='day', y='count', title="Day"))\
.write_html("static/plots/19.html")

style(px.line(df.groupby('day_of_year').size().reset_index(name='count'),
              x='day_of_year', y='count', title="Day of Year"))\
.write_html("static/plots/20.html")

# ===============================
# 🔷 21–30 SPEED
# ===============================

style(px.bar(df, x='speed_category', color='cme_type',
             title="Speed vs Type"))\
.write_html("static/plots/21.html")

style(px.box(df, x='cme_type', y='speed', title="Speed vs Type Box"))\
.write_html("static/plots/22.html")

style(px.histogram(df, x='speed', color='cme_type',
                   title="Speed Type Dist"))\
.write_html("static/plots/23.html")

style(px.scatter(df, x='speed', y='latitude', title="Speed vs Lat"))\
.write_html("static/plots/24.html")

style(px.scatter(df, x='speed', y='longitude', title="Speed vs Lon"))\
.write_html("static/plots/25.html")

style(px.box(df, x='speed_category', y='half_angle',
             title="Angle vs Speed Cat"))\
.write_html("static/plots/26.html")

style(px.histogram(df, x='half_angle', title="Half Angle"))\
.write_html("static/plots/27.html")

style(px.box(df, y='half_angle', title="Half Angle Box"))\
.write_html("static/plots/28.html")

style(px.scatter(df, x='half_angle', y='latitude',
                 title="Angle vs Lat"))\
.write_html("static/plots/29.html")

style(px.scatter(df, x='half_angle', y='longitude',
                 title="Angle vs Lon"))\
.write_html("static/plots/30.html")

# ===============================
# 🔷 31–40 GEO
# ===============================

style(px.scatter(df, x='longitude', y='latitude',
                 title="Lat vs Lon"))\
.write_html("static/plots/31.html")

style(px.histogram(df, x='latitude', title="Lat Dist"))\
.write_html("static/plots/32.html")

style(px.histogram(df, x='longitude', title="Lon Dist"))\
.write_html("static/plots/33.html")

style(px.box(df, y='latitude', title="Lat Box"))\
.write_html("static/plots/34.html")

style(px.box(df, y='longitude', title="Lon Box"))\
.write_html("static/plots/35.html")

style(px.scatter(df, x='latitude', y='speed',
                 title="Lat vs Speed"))\
.write_html("static/plots/36.html")

style(px.scatter(df, x='longitude', y='speed',
                 title="Lon vs Speed"))\
.write_html("static/plots/37.html")

style(px.histogram(df, x='direction', title="Direction"))\
.write_html("static/plots/38.html")

style(px.bar(df.groupby('direction_bin').size().reset_index(name='count'),
             x='direction_bin', y='count', title="Direction Bin"))\
.write_html("static/plots/39.html")

style(px.scatter(df, x='direction', y='speed',
                 title="Direction vs Speed"))\
.write_html("static/plots/40.html")

# ===============================
# 🔷 41–50 IMPACT
# ===============================

style(px.pie(df, names='potentially_geoeffective',
             title="Geo Impact", hole=0.4))\
.write_html("static/plots/41.html")

style(px.box(df, x='potentially_geoeffective', y='speed',
             title="Speed vs Geo"))\
.write_html("static/plots/42.html")

style(px.bar(df, x='cme_type', color='potentially_geoeffective',
             title="Type vs Geo"))\
.write_html("static/plots/43.html")

style(px.bar(df, x='event_type', color='potentially_geoeffective',
             title="Event vs Geo"))\
.write_html("static/plots/44.html")

style(px.histogram(df, x='speed', color='potentially_geoeffective',
                   title="Speed Geo Dist"))\
.write_html("static/plots/45.html")

style(px.box(df, y='half_angle', color='potentially_geoeffective',
             title="Angle vs Geo"))\
.write_html("static/plots/46.html")

style(px.scatter(df, x='speed', y='latitude',
                 color='potentially_geoeffective',
                 title="Geo Scatter"))\
.write_html("static/plots/47.html")

style(px.bar(df, x='measurement_technique',
             color='potentially_geoeffective',
             title="Tech vs Geo"))\
.write_html("static/plots/48.html")

style(px.bar(df, x='catalog',
             color='potentially_geoeffective',
             title="Catalog vs Geo"))\
.write_html("static/plots/49.html")

style(px.bar(df, x='source_location',
             color='potentially_geoeffective',
             title="Location vs Geo"))\
.write_html("static/plots/50.html")

print("🔥 ALL 50+ GRAPHS GENERATED SUCCESSFULLY")
```

```python
def style(fig):
    fig.update_layout(
        template="plotly_dark",
        paper_bgcolor="#000814",
        plot_bgcolor="#001d3d",
        font=dict(color="white", size=12),

        title=dict(
            x=0.5,
            font=dict(size=18, color="#00f7ff")
        ),

        margin=dict(l=20, r=20, t=40, b=20),

        hovermode="x unified"
    )

    fig.update_traces(
        marker=dict(line=dict(width=0)),
        opacity=0.9
    )

    return fig
```
